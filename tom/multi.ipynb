```python

import pandas as pd
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, f1_score, classification_report

# Load dataset
data = fetch_20newsgroups(subset='all')

X = data.data
y = data.target

# Convert text into numerical values
v = CountVectorizer()
X = v.fit_transform(X)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Multinomial Naive Bayes
m = MultinomialNB()
m.fit(X_train, y_train)
p1 = m.predict(X_test)

# Bernoulli Naive Bayes
b = BernoulliNB()
b.fit(X_train, y_train)
p2 = b.predict(X_test)

# Calculate metrics
for name, p in [('Multinomial NB', p1), ('Bernoulli NB', p2)]:
    print("\n", name)
    print("Accuracy:", accuracy_score(y_test, p))
    print("F1 Score:", f1_score(y_test, p, average='weighted'))

# Comparison table
df = pd.DataFrame({
    'Model': ['Multinomial NB', 'Bernoulli NB'],
    'Accuracy': [accuracy_score(y_test, p1), accuracy_score(y_test, p2)],
    'F1 Score': [
        f1_score(y_test, p1, average='weighted'),
        f1_score(y_test, p2, average='weighted')
    ]
})

print("\nComparison")
print(df)

# Classification report
print("\nClassification Report")
print(classification_report(y_test, p1, target_names=data.target_names))
```
