```python
import pandas as pd
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score

data = load_digits()

x = data.data
y = data.target

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=43
)

x_train = x_train / 255.0
x_test = x_test / 255.0

linear_model = SVC(kernel='linear')
linear_model.fit(x_train, y_train)
p1 = linear_model.predict(x_test)

poly_model = SVC(kernel='poly', degree=3)
poly_model.fit(x_train, y_train)
p2 = poly_model.predict(x_test)

rbf_model = SVC(kernel='rbf')
rbf_model.fit(x_train, y_train)
p3 = rbf_model.predict(x_test)

print("\n Linear SVM")
print("------------------------------------------")
print("Accuracy :", accuracy_score(y_test, p1))
print("Precision:", precision_score(y_test, p1, average='weighted'))
print("Recall   :", recall_score(y_test, p1, average='weighted'))
print("F1 Score :", f1_score(y_test, p1, average='weighted'))

print("\n Polynomial SVM")
print("------------------------------------------")
print("Accuracy :", accuracy_score(y_test, p2))
print("Precision:", precision_score(y_test, p2, average='weighted'))
print("Recall   :", recall_score(y_test, p2, average='weighted'))
print("F1 Score :", f1_score(y_test, p2, average='weighted'))

print("\n RBF SVM")
print("------------------------------------------")
print("Accuracy :", accuracy_score(y_test, p3))
print("Precision:", precision_score(y_test, p3, average='weighted'))
print("Recall   :", recall_score(y_test, p3, average='weighted'))
print("F1 Score :", f1_score(y_test, p3, average='weighted'))
```
