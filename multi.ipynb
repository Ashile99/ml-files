```python
import pandas as pd
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, f1_score, classification_report

data = ["comp.graphics", "rec.sport.baseball",
        "sci.space", "talk.politics.misc"]

newsgroups = fetch_20newsgroups(
    subset="all",
    categories=data,
    remove=("headers", "footers", "quotes")
)

X = newsgroups.data
y = newsgroups.target

v = CountVectorizer()
X = v.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

m = MultinomialNB()
m.fit(X_train, y_train)
p1 = m.predict(X_test)

b = BernoulliNB()
b.fit(X_train, y_train)
p2 = b.predict(X_test)

print("Multinomial Naive Bayes")
print("Accuracy :", accuracy_score(y_test, p1))
print("F1 Score :", f1_score(y_test, p1, average="weighted"))

print("\nBernoulli Naive Bayes")
print("Accuracy :", accuracy_score(y_test, p2))
print("F1 Score :", f1_score(y_test, p2, average="weighted"))

df = pd.DataFrame({
    "Model": ["Multinomial NB", "Bernoulli NB"],
    "Accuracy": [
        accuracy_score(y_test, p1),
        accuracy_score(y_test, p2)
    ],
    "F1 Score": [
        f1_score(y_test, p1, average="weighted"),
        f1_score(y_test, p2, average="weighted")
    ]
})

print("\nComparison")
print(df)

print("\nMultinomial NB Report")
print(classification_report(
    y_test, p1,
    target_names=newsgroups.target_names
))

print("\nBernoulli NB Report")
print(classification_report(
    y_test, p2,
    target_names=newsgroups.target_names
))
```
